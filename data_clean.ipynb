```python
import pandas as pd
df = pd.read_excel("data/cardiac_patient_data.xlsx")
df.head()
```

```python
# print(df['Data przyjęcia'].equals(df['Data przyjęcia.1']))
df.drop(columns=['Data przyjęcia.1'], inplace=True)     # duplicated column

df['data zgonu lub konca obserwacji'] = df['data zgonu lub konca obserwacji'].astype(str).str.replace('31.04.2020', '2020-04-30 00:00:00')   
df['data zgonu lub konca obserwacji'] = pd.to_datetime(df['data zgonu lub konca obserwacji'], format='%Y-%m-%d %H:%M:%S')
# invalid date, ask for confirmation

df['Unnamed: 5'] = df['data zgonu lub konca obserwacji'] - df['Data przyjęcia']
df.rename(columns={'Unnamed: 5': 'Time of observation'}, inplace=True)
# correction of column corupted by invalid date
```

```python
df.convert_dtypes()
df.to_csv("data/cardiac_patient_data.csv", index=False)
df.head()
```
